```python
def createUser(firstName: str, lastName: str, age: int | None = None) -> dict[str, str | None | int]:      # -> dict means it should return a dict; [] means what the keys and/or values should return 
    email = f"{firstName.lower()}_{lastName.lower()}@gmail.com"
    
    return{
        "First Name": firstName,
        "Last Name": lastName,
        "Email": email,
        "Age": age
    }

user1 = createUser("Hardeep", "Ghotra", 20)
user2 = createUser("Hardeep", "Singh")

print(user1)
print(user2)
```

```python
# Type Alias
type User = dict[str, str | None | int]

def createUser(firstName: str, lastName: str, age: int | None = None) -> User:
    email = f"{firstName.lower()}_{lastName.lower()}@gmail.com"
    
    return{
        "First Name": firstName,
        "Last Name": lastName,
        "Email": email,
        "Age": age
    }

user1 = createUser("Hardeep", "Ghotra", 20)
user2 = createUser("Hardeep", "Singh")

print(user1)
print(user2)
```

```python
from typing import TypedDict


class User(TypedDict):
    firstName: str
    lastName: str
    email: str
    age: int | None

def createUser(firstName: str, lastName: str, age: int | None = None) -> User:
    email = f"{firstName.lower()}_{lastName.lower()}@gmail.com"
    
    return{
        "First Name": firstName,
        "Last Name": lastName,
        "Email": email,
        "Age": age
    }
```

```python
from dataclasses import dataclass


@dataclass
class User:
    firstName: str
    lastName: str
    email: str
    age: int | None = None

def createUser(firstName: str, lastName: str, age: int | None = None) -> User:
    email = f"{firstName.lower()}_{lastName.lower()}@gmail.com"
    
    return User( 
        firstName = firstName,
        lastName =  lastName,
        email =  email,
        age = age
    )

user1 = createUser("Hardeep", "Ghotra", 20)
user2 = createUser("Hardeep", "Singh")

print(user1)
print(user2)
```

```python
# Generic types 
import random


def randomChoice(items: list[User]) -> User:
    return random.choice(items)

users = [user1, user2]
rando_user = randomChoice(users)
print(rando_user)

emails = [user.email for user in users]
rando_email = randomChoice(emails)
print(rando_email)
```
