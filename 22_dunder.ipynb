```python
class Trial:
    def __init__(self, name):
        self.name = name
    
    def __str__(self):
        return f"Hello my name is {self.name}"
    
obj = Trial("Hardeep")
print(obj)
```

```python
class Numbers:
    def __init__(self, num):
        self.num = num
    
    def __add__(self, other):
        return self.num + other.num

a = Numbers(20)
b = Numbers(30)

a + b
```
