```python
class Animal:
    def speak(cls):
        print("Animal will not speak")

class Human:
    def speak(cls):
        print("Humans will speak")
        
obj1 = Animal()
obj2 = Human()

obj1.speak()
obj2.speak()
```

```python
# Method Overriding: [Need Inheritence]

class Animal:
    a = 11
    def __init__(self, name):
        self.name = name
        
    def details(self):
        print("Hello", self.name)

class Humans(Animal):
    b = 17
    def details(self):
        print(f"You are {self.name}")

obj = Humans("Hardeep")

print(obj.a)
print(obj.b)
obj.details() # Only the second class(child class) for which the object has been created will run, called as method overriding
```
