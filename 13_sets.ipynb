```python
# To see the hash value:
s = "Hello"
hash(s)
```

```python
# Can not access items using indexes
# Only way for for loop in sets:

s = {1, 2, 3, 4}

for i in s:
    print(s)
```

## Math operations for two sets

```python
s1 = {10, 20, 30, 40}
s2 = {30, 40, 50, 60}

print(s1.difference(s2)) # Gives only unique elements of s1 in comparasible to s2
print(s1 - s2)
```

```python
s2 -= s1 # Updates s2 and remove duplicate elements
s2
```

```python
print(s1.intersection(s2)) # Gives only common elements
print(s1 & s2)
```

```python
s1 &= s2 # Updates s1 and remove unique elements
s1
```

```python
s3 = {30, 40}
# S3 is a subset of s2, but s2 is not a subset of s3

print(s3 <= s2)
print(s2 <= s3)

print("\n")

# S2 is a superset of S3, but S3 is not a superset of s2
print(s3 >= s2)
print(s2 >= s3)
```

```python
s1 = {10, 20, 30, 40}
s2 = {30, 40, 50, 60}

# To get a new set with only unique elements:
print(s1.symmetric_difference(s2))
print(s1 ^ s2)
```

```python
s1 ^= s2
s1
```

```python
# To get all elements
print(s1.union(s2))
print(s1 | s2)
```
