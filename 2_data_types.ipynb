```python
# Int --> -infinity to +infinity
data_int = 23
print(type(data_int))
```

```python
# Float --> numbers with decimals and fractions
data_float = 23.5
data_float_fraction = 2/3
print(type(data_float))
print(type(data_float_fraction))
```

```python
# Boolean --> True or False
data_bool = True
print(type(data_bool))
```

```python
# String --> Text (any text)
data_string = "Hello my name is Hardeep"
print(type(data_string))
```
