```python
class BagFactory:
    def __init__(self, materials, zips, pockets):
        self.materials = materials
        self.zips = zips
        self.pockets = pockets
    
    def details(self):
        print(f"Your bag details are: {self.materials}, {self.pockets}, {self.zips}")

class Reebok(BagFactory):
    def __init__(self, materials, zips, pockets, color):
        super().__init__(materials, zips, pockets)   # Super keyword is used to get access to parent class's attributes and methods
        self.color = color
    
    def details(self):
        print("Color: ", self.color)
        return super().details()

bag1 = BagFactory("leather", 4, 5)
bag2 = ("polyester", 8, 12, "red")

bag1.details()
```
