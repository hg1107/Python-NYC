```python
# Key: Value pair
# Keys should be unique

d = {1: 10,
     2: 20,
     3: 30,
     4: 40,
     5: 50}

# Accessible using keys and not indexes
print(d[1])
```

## Vanilla Python

```python
# Creating new key-value pair
d[6] = 60
print(d)

# Updating a value:
d[5] = 55
print(d)
```

## Methods Approach

```python
d = {1: 10,
     2: 20,
     3: 30,
     4: 40,
     5: 50}

print("Original: ", d, "\n")

q = d.fromkeys([10, 20], 50)
print(q, "\n")

# Return value of specified key
print(d.get(1), "\n")

# List containing tuples of key and value
print(d.items(), "\n")

# List containing keys
print(d.keys(), "\n")

# For values:
print(d.values(), "\n")

# Deleting item using key
print(d.pop(1))
print(d, "\n")

# Remove last inserted
print(d.popitem())
print(d, "\n")

# Returns the value of specified key, and if the key is not present, then it creates a new key
print(d.setdefault(3))
print(d.setdefault(10, 100))
print(d, "\n")

# Update:
d.update({10: 1000})
d.update({7: 70})
print(d)
```

```python
# 1. Merge two dictionaries into one:
d1 = {1: 10, 2: 20, 3: 30}
d2 = {4: 40, 5: 50}

d1.update(d2)
print(d1)
```

```python
# Sum all values in a dictionary:
d = {"a": 10, "b": 20, "c": 30}

lst = d.values()
sum(lst)
```

```python
# Count the frequency of each element in a list using a dictionary:
lst = [1, 2, 1, 3, 2, 1]

d = {}
for i in lst:
    if i in d.keys():
        d[i] = d[i] + 1
    else:
        d[i] = 1

print(d)
```

```python
# Combine two dicts, adding values to common keys:
d1 = {"a": 5, "b": 3}
d2 = {"b": 4, "c": 2}

for i in d2:
    if i in d1.keys():
        d1[i] = d1[i] + d2[i] 
    else:
        d1[i] = d2[i]

print(d1)
```
