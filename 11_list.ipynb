```python
'''

Collection of multiple data types
list = []
Ordered
Indexed
Mutable
Duplications allowed

'''
```

```python
list = [1, 'Hello', True, 2.5]
```

```python
# Complete list
print(list)
```

```python
# Indexing
print(list[1])
```

```python
# Mutable
list[-1] = 3.14
print(list)
```

```python
# Traversing (loops) on lists
a = [10, 20, 30, 40, 50]
```

```python
# 1. Traversing on values:
for i in a:
    print(i, end = " ")
```

```python
# 2. Traversing on indexes
for i in range(len(a)):
    print(f"{i}: {a[i]}")
```

```python
# List Methods:
# CRUD --> Create, Read, Update, Delete

# To get the available methods for a certain datatype:
print(dir(list))
```

```python
# Append
a = [10, 20, 30, 40]
print(a)

a.append(60)
a.append("Hello")
print(a)
```

```python
a.insert(1, "Inserted")
print(a)
```

```python
# Pop
a.pop()
print(a)

a.pop(1)
print(a)
```

```python
# Remove:
a.remove(60)
print("\n", a)
```
