```python
# Traversing (loops) on lists
a = [10, 20, 30, 40, 50]
```

```python
# 1. Traversing on values:
for i in a:
    print(i, end = " ")
```

```python
# 2. Traversing on indexes
for i in range(len(a)):
    print(f"{i}: {a[i]}")
```

```python
# List Methods:
# CRUD --> Create, Read, Update, Delete

# To get the available methods for a certain datatype:
print(dir(list))
```

```python
# Append
a = [10, 20, 30, 40]
print(a)

a.append(60)
a.append("Hello")
print(a)
```

```python
a.insert(1, "Inserted")
print(a)
```

```python
# Pop
a.pop()
print(a)

a.pop(1)
print(a)
```

```python
# Remove:
a.remove(60)
print("\n", a)
```

```python
# Print all positives and negatives elements seperately

lst = [3, -1, 4, -5, 9]
pos_lst = []
neg_lst = []

for i in range(len(lst)):
    if lst[i] > 0:
        pos_lst.append(lst[i])
    else:
        neg_lst.append(lst[i])

print("Positive elements: ", pos_lst)
print("Negative elements: ", neg_lst)
```

```python
# FInd average of all list elements:
avg = [10, 20, 30, 40]
mean = sum(avg) / len(avg)
print("Average of all the elements of list: ", mean)
```

```python
# Find the greatest element and print its index:
idx = [4, 8, 2, 9, 1, 9, 9, 9, 9, 9]
maxm = max(idx)
for i in range(len(idx)):
    if maxm == idx[i]:
        max_idx = i
        break

print(f"Maximum value is {maxm} at index {max_idx}")
```

```python
# Find the second greatest element:
for i in range(idx[i]):
    if maxm in idx:
        idx.remove(maxm)
print("Second greatest element: ", max(idx))
```
