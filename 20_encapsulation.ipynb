```python
class Factory:
    name = "Kia"  # Public class attribute'
    _old = 12     # Protected member
    __year = 2002 # Private Member
    def __init__(self, type, tyre, color):
        self.type = type    # Public object attribute
        self.__tyre = tyre
        self.color = color
    
    def __details(self):
        print("Hello")
        
class Hello(Factory):
    print(Factory.name)
    # print(Factory.__year) Error
        
obj = Factory("Sedan", "MRF", "Black")

obj.name = "Suzuki"
obj._old = 11

# print(obj.__year)     Will give error
# obj.__details()       Will give error
```
