```python
class Car:
    a = 12 # Attribute
    
    def hello(): # Method
        print("Helloo")

print(Car.a) # Accessing attributes
Car.hello() # Accesing methods
```

```python
class Bags:
    name = "Factory"
    
    def details(self):
        print("Hello this is a company who creates bags")

skybag = Bags()     #Object
reebok = Bags()

print(skybag.name)
print(reebok.name)

skybag.details()
```

## Constructor
Method that runs automatically when a class is called

```python
class Bags:
    def __init__(self):
        print("How are you")

Bags()
```

```python
class Bags:
    def __init__(self, material, zips, pockets):
        self.material = material
        self.zips = zips
        self.pockets = pockets

reebok = Bags("Leather", 3, 2)
skybag = Bags("Polyester", 4, 6)

print(reebok.material)
print(skybag.material)
```

## Types of Methods

```python
class Methods:
    def hello(self):    # Instance/Object method. Captures the location of object
        print("YOO")
    
    @classmethod
    def details(cls):   # Class method. Captures the location of class
        print("How are you")
    
    @staticmethod
    def speak():    # Static method and does not target any location
        print("Hello")
```
